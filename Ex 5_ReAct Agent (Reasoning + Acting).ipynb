```python
import getpass
from openai import OpenAI

# 1. Get Azure AI Foundry API key securely
api_key = getpass.getpass("Enter your Azure AI Foundry API key: ")

# 2. Configure OpenAI client pointing to your Azure AI Foundry endpoint
client = OpenAI(
    api_key=api_key.strip(),
    base_url="https://ganapathy-0333-resource.services.ai.azure.com/openai/v1/"
)

# Mock search function
def search_web(query):
    return f"Results for '{query}': India won 2024 T20 World Cup"

# ReAct Agent loop
def react_agent(goal, max_steps=5):
    history = []

    for step in range(max_steps):
        prompt = f"""You are a ReAct agent. Pattern:
Thought: Think about what to do
Action: [search/answer]
Action Input: Input for the action

Goal: {goal}
Previous steps: {history}

Next step?"""

        # Call Azure AI Foundry API
        response = client.chat.completions.create(
            model="gpt-4.1-mini",
            messages=[{"role": "user", "content": prompt}]
        )

        step_text = response.choices[0].message.content.strip()
        history.append(f"Step {step + 1}: {step_text}")
        print(f"\n--- Step {step + 1} ---\n{step_text}")

        if "Action: search" in step_text:
            query = step_text.split("Action Input:")[-1].strip()
            result = search_web(query)
            history.append(f"Observation: {result}")
            print(f"Observation: {result}")
        elif "Action: answer" in step_text:
            print("\n✅ Task completed!")
            break

    return history

# Run the agent
if __name__ == "__main__":
    react_agent("Who won the 2024 T20 World Cup?")
```
